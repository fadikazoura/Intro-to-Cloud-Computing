```python
from google.colab import drive
drive.mount('/content/drive')

import json
FILE = '/content/drive/MyDrive/students.json'
```

```python
students = {
    "students": [
        {
            "first_name": "Fadi",
            "last_name": "Kazoura",
            "email": "fadi.kazoura@e.braude.ac.il",
            "courses": ["אבטחת מידע וקריפטולוגיה", "מבוא לבינה מלאכותית", "web", "מבוא ל מחשוב ענן", "כריית נתונים ומערכות לומדות"],
            "link": "https://w3.braude.ac.il/",
            "favorite_program": ""
        },
        {
            "first_name": "Rani",
            "last_name": "Renawe",
            "email": "rani.renawe@e.braude.ac.il",
            "courses": ["אבטחת מידע וקריפטולוגיה", "מבוא לבינה מלאכותית", "web", "מבוא ל מחשוב ענן", "כריית נתונים ומערכות לומדות"],
            "link": "https://www.youtube.com/",
            "favorite_program": ""
        }
    ]
}

with open(FILE, 'w', encoding='utf-8') as f:
    json.dump(students, f, indent=4, ensure_ascii=False)
print("students.json created in Drive")
```

```python
import ipywidgets as widgets
from IPython.display import display

def load_data():
    with open(FILE, encoding='utf-8') as f:
        return json.load(f)

def save_data(d):
    with open(FILE, 'w', encoding='utf-8') as f:
        json.dump(d, f, indent=4, ensure_ascii=False)

data = load_data()
students = data["students"]

style = {'description_width': '140px'}
lay = widgets.Layout(width='460px')

title = widgets.HTML("<h2 style='color:#1a73e8;margin:0'>🎓 Group Students</h2><hr>")

dropdown = widgets.Dropdown(
    options=[(f"{s['first_name']} {s['last_name']}", i) for i, s in enumerate(students)],
    value=None, description='Choose student:', style=style, layout=lay)

first   = widgets.Text(description='First name:', disabled=True, style=style, layout=lay)
last    = widgets.Text(description='Last name:',  disabled=True, style=style, layout=lay)
email   = widgets.Text(description='Braude mail:', disabled=True, style=style, layout=lay)
courses = widgets.Textarea(description='Courses:', disabled=True, style=style,
                           layout=widgets.Layout(width='460px', height='90px'))
link    = widgets.Text(description='Interesting link:', disabled=True, style=style, layout=lay)
link_html = widgets.HTML()

fav = widgets.Text(description='Favorite program:', placeholder='Type a program...',
                   style=style, layout=lay)
btn = widgets.Button(description='Update', button_style='success', icon='save')
msg = widgets.HTML()

def on_select(change):
    i = change['new']
    if i is None:
        return
    s = students[i]
    first.value   = s['first_name']
    last.value    = s['last_name']
    email.value   = s['email']
    courses.value = '\n'.join(s['courses'])
    link.value    = s['link']
    link_html.value = f"<a href='{s['link']}' target='_blank'>🔗 Open link</a>"
    fav.value     = s.get('favorite_program', '')
    msg.value     = ''

def on_update(b):
    i = dropdown.value
    if i is None:
        msg.value = "<b style='color:red'>Choose a student first!</b>"
        return
    students[i]['favorite_program'] = fav.value.strip()
    save_data(data)          # writes back to students.json in Drive
    msg.value = f"<b style='color:green'>✔ Updated {students[i]['first_name']}'s favorite program</b>"

dropdown.observe(on_select, names='value')
btn.on_click(on_update)

form = widgets.VBox(
    [title, dropdown, first, last, email, courses, link, link_html,
     widgets.HTML("<hr>"), fav, btn, msg],
    layout=widgets.Layout(border='2px solid #1a73e8', padding='15px', width='520px'))
display(form)
```
